```python
import pandas as pd

df = pd.read_excel(
    "/content/curriculum_industry_skill_gap_dataset_800.xlsx"
)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

display(df.head())
```

```python
print("Dataset Information")
print("===================")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nMissing values:")
print(df.isnull().sum())

print("\nData types:")
print(df.dtypes)
```

```python
feature_df = df[
    [
        "Student_ID",
        "Programming",
        "DSA",
        "DBMS",
        "Computer_Networks",
        "Operating_Systems",
        "Web_Development",
        "Data_Analytics",
        "AI_ML",
        "Cloud_Computing",
        "Cybersecurity",
        "Problem_Solving",
        "Aptitude",
        "Communication",
        "Teamwork",
        "Git_GitHub"
    ]
].copy()

print("Feature dataset created!")
print("Shape:", feature_df.shape)

display(feature_df.head())
```

```python
feature_df["technical_skill_average"] = feature_df[
    [
        "Programming",
        "DSA",
        "DBMS",
        "Computer_Networks",
        "Operating_Systems",
        "Web_Development",
        "Data_Analytics",
        "AI_ML",
        "Cloud_Computing",
        "Cybersecurity"
    ]
].mean(axis=1)

feature_df["programming_problem_solving"] = feature_df[
    ["Programming", "DSA", "Problem_Solving"]
].mean(axis=1)

feature_df["data_ai_average"] = feature_df[
    ["Data_Analytics", "AI_ML"]
].mean(axis=1)

feature_df["cloud_security_average"] = feature_df[
    ["Cloud_Computing", "Cybersecurity"]
].mean(axis=1)

feature_df["soft_skill_average"] = feature_df[
    ["Communication", "Teamwork"]
].mean(axis=1)

feature_df["overall_skill_average"] = feature_df[
    [
        "Programming",
        "DSA",
        "DBMS",
        "Computer_Networks",
        "Operating_Systems",
        "Web_Development",
        "Data_Analytics",
        "AI_ML",
        "Cloud_Computing",
        "Cybersecurity",
        "Problem_Solving",
        "Aptitude",
        "Communication",
        "Teamwork",
        "Git_GitHub"
    ]
].mean(axis=1)

print("Feature engineering completed!")
display(feature_df.head())
```

```python
# Add industry-related features and timestamp

feature_df["overall_industry_gap"] = df["Overall_Industry_Gap"]

feature_df["curriculum_alignment_score"] = df[
    "Curriculum_Alignment_Score"
]

feature_df["industry_alignment_score"] = df[
    "Industry_Alignment_Score"
]

# Feast requires an entity column called student_id
feature_df = feature_df.rename(
    columns={"Student_ID": "student_id"}
)

# Timestamp for Feast
feature_df["event_timestamp"] = df["event_timestamp"]

print("Feast dataset prepared!")
print("Shape:", feature_df.shape)

display(feature_df.head())
```

```python
%%writefile /content/feature_repo/features.py

from datetime import timedelta

from feast import Entity, FeatureView, Field, FileSource
from feast.types import Float32


# 1. ENTITY

student = Entity(
    name="student_id",
    join_keys=["student_id"],
    description="Unique identifier for each CSE student"
)


# 2. DATA SOURCE

student_features_source = FileSource(
    name="student_features_source",
    path="data/student_features.parquet",
    timestamp_field="event_timestamp"
)


# 3. FEATURE VIEW

student_skill_features = FeatureView(
    name="student_skill_features",
    entities=[student],
    ttl=timedelta(days=365),

    schema=[
        Field(name="Programming", dtype=Float32),
        Field(name="DSA", dtype=Float32),
        Field(name="DBMS", dtype=Float32),
        Field(name="Computer_Networks", dtype=Float32),
        Field(name="Operating_Systems", dtype=Float32),
        Field(name="Web_Development", dtype=Float32),
        Field(name="Data_Analytics", dtype=Float32),
        Field(name="AI_ML", dtype=Float32),
        Field(name="Cloud_Computing", dtype=Float32),
        Field(name="Cybersecurity", dtype=Float32),
        Field(name="Problem_Solving", dtype=Float32),
        Field(name="Aptitude", dtype=Float32),
        Field(name="Communication", dtype=Float32),
        Field(name="Teamwork", dtype=Float32),
        Field(name="Git_GitHub", dtype=Float32),

        Field(name="technical_skill_average", dtype=Float32),
        Field(name="programming_problem_solving", dtype=Float32),
        Field(name="data_ai_average", dtype=Float32),
        Field(name="cloud_security_average", dtype=Float32),
        Field(name="soft_skill_average", dtype=Float32),
        Field(name="overall_skill_average", dtype=Float32),
        Field(name="overall_industry_gap", dtype=Float32),
        Field(name="curriculum_alignment_score", dtype=Float32),
        Field(name="industry_alignment_score", dtype=Float32),
    ],

    source=student_features_source,
)

print("features.py created successfully!")
```

```python
from feast import FeatureStore

store = FeatureStore(
    repo_path="/content/feature_repo"
)

feature_refs = [
    "student_skill_features:Programming",
    "student_skill_features:DSA",
    "student_skill_features:DBMS",
    "student_skill_features:Computer_Networks",
    "student_skill_features:Operating_Systems",
    "student_skill_features:Web_Development",
    "student_skill_features:Data_Analytics",
    "student_skill_features:AI_ML",
    "student_skill_features:Cloud_Computing",
    "student_skill_features:Cybersecurity",
    "student_skill_features:Problem_Solving",
    "student_skill_features:Aptitude",
    "student_skill_features:Communication",
    "student_skill_features:Teamwork",
    "student_skill_features:Git_GitHub",
    "student_skill_features:technical_skill_average",
    "student_skill_features:programming_problem_solving",
    "student_skill_features:data_ai_average",
    "student_skill_features:cloud_security_average",
    "student_skill_features:soft_skill_average",
    "student_skill_features:overall_skill_average",
    "student_skill_features:overall_industry_gap",
    "student_skill_features:curriculum_alignment_score",
    "student_skill_features:industry_alignment_score"
]

entity_df = feature_df[
    ["student_id", "event_timestamp"]
].copy()

historical_df = store.get_historical_features(
    entity_df=entity_df,
    features=feature_refs
).to_df()

print("✅ HISTORICAL FEATURE RETRIEVAL SUCCESSFUL")
print("Shape:", historical_df.shape)

display(historical_df.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

model_features = [
    "Programming",
    "DSA",
    "DBMS",
    "Computer_Networks",
    "Operating_Systems",
    "Web_Development",
    "Data_Analytics",
    "AI_ML",
    "Cloud_Computing",
    "Cybersecurity",
    "Problem_Solving",
    "Aptitude",
    "Communication",
    "Teamwork",
    "Git_GitHub",
    "technical_skill_average",
    "programming_problem_solving",
    "data_ai_average",
    "cloud_security_average",
    "soft_skill_average",
    "overall_skill_average",
    "overall_industry_gap",
    "curriculum_alignment_score",
    "industry_alignment_score"
]

X = historical_df[model_features]
y = df["Skill_Gap_Category"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("================================")
print("MODEL RESULTS")
print("================================")
print("Accuracy:", accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
from datetime import timedelta

start_time = feature_df["event_timestamp"].min().to_pydatetime()

end_time = (
    feature_df["event_timestamp"].max()
    + timedelta(hours=1)
).to_pydatetime()

print("Materializing from:", start_time)
print("Materializing until:", end_time)

store.materialize(
    start_time,
    end_time
)

print("================================")
print("MATERIALIZATION SUCCESSFUL")
print("================================")
```

```python
online_features = store.get_online_features(
    features=feature_refs,
    entity_rows=[
        {"student_id": "S0001"}
    ]
).to_dict()

print("================================")
print("ONLINE FEATURE OUTPUT")
print("================================")

for key, value in online_features.items():
    print(key, ":", value)
```

```python
# Convert Feast online features into a DataFrame

online_row = {}

for key, value in online_features.items():
    if isinstance(value, list):
        online_row[key] = value[0]
    else:
        online_row[key] = value

online_df = pd.DataFrame([online_row])

# Prepare model input
prediction_input = online_df.reindex(
    columns=model_features
)

# Make prediction
prediction = model.predict(prediction_input)

print("================================")
print("FINAL PREDICTION")
print("================================")
print("Student ID: S0001")
print("Predicted Skill Gap:", prediction[0])
print("================================")
```
